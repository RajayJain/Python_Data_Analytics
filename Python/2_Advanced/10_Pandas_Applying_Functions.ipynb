```python
import ast

def clean_list(skill_list):
    # Only run literal_eval if the row is actually a string
    if isinstance(skill_list, str):
        return ast.literal_eval(skill_list)
    
    # If it's already a list or a NaN (float), just return it as-is
    return skill_list

df['job_skills'] = df['job_skills'].apply(clean_list)
```
