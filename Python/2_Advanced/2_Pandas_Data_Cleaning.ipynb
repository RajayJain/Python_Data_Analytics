```python
import pandas as pd
import matplotlib.pyplot as plt
from datasets import load_dataset
#Load the dataset
dataset = load_dataset("lukebarousse/data_jobs")
df = dataset["train"].to_pandas()
#Data Cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
df.loc[:5, 'salary_rate' : 'salary_hour_avg']
```

```python
median_salary_year = df['salary_year_avg'].median()
```

```python
median_salary_hour = df['salary_hour_avg'].median()
```

```python
df_filled = df
df_filled['salary_year_avg'] = df_filled['salary_year_avg'].fillna(median_salary_year)
df_filled['salary_hour_avg'] = df_filled['salary_hour_avg'].fillna(median_salary_hour)
```

```python
df_filled.loc[:5, 'salary_year_avg' : 'salary_hour_avg']
```

```python
df_unique = df_filled
df_unique = df_unique.drop_duplicates()
print('Length before dropping duplicates:', len(df_filled))
print('Length after dropping duplicates:', len(df_unique))
print('Number of duplicates dropped:', len(df_filled) - len(df_unique))
```

```python
df_unique = df_unique.drop_duplicates(subset=['job_title', 'company_name'])
print('Length before dropping duplicates:', len(df_filled))
print('Length after dropping duplicates:', len(df_unique))
print('Number of duplicates dropped:', len(df_filled) - len(df_unique))
```
