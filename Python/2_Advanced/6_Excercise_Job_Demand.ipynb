```python
import pandas as pd
import matplotlib.pyplot as plt
from datasets import load_dataset
#Load the dataset
dataset = load_dataset("lukebarousse/data_jobs")
df = dataset["train"].to_pandas()
#Data Cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
df_US = df[df['job_country'] == 'United States'].copy()
df_US
```

```python
df_US['job_posted_month'] = df_US['job_posted_date'].dt.strftime('%B')
df_US
```

```python
df_US_pivot = df_US.pivot_table(index='job_posted_month', aggfunc='size', columns = 'job_title_short')
```

```python
df_US_pivot.reset_index(inplace=True)
df_US_pivot['month_number'] = pd.to_datetime(df_US_pivot['job_posted_month'], format='%B').dt.month
df_US_pivot['month_number']
```

```python
df_US_pivot.sort_values('month_number', inplace=True)
df_US_pivot.set_index('job_posted_month', inplace=True)
```

```python
df_US_pivot
```

```python
df_US_pivot.drop(columns='month_number', inplace=True)
```

```python
df_US_pivot
```

```python
df_US_pivot.plot(kind='line', figsize=(10, 5))
plt.title('Job Demand in the United States by Month')
```

```python
top_3 = df_US['job_title_short'].value_counts().head(3).index
top_3 = top_3.to_list()
```

```python
df_US_pivot[top_3].plot(kind='line', figsize=(10, 5))
plt.title('Top 3 Job Demand in the United States by Month')
plt.xlabel('2024')
plt.ylabel('Number of Job Postings')
plt.legend(title='Job Title')
plt.show()
```
